```python
import pandas as pd

# Load lookup table used for adm2_code assignment
# (Path differs by repository/package layout)
lookup_path = 'inst/extdata/lookup_district_code.csv'

# Example safe loader pattern
try:
    lookup = pd.read_csv(lookup_path)
    lookup.head()
except FileNotFoundError:
    print('Update lookup_path to your local clone/package path before running.')
```

## Post-processing template


```python
def assign_adm2_code_by_name(post_df: pd.DataFrame, lookup: pd.DataFrame) -> pd.DataFrame:
    # post_df is expected to contain: adm2kr, year
    merged = post_df.merge(
        lookup,
        left_on='adm2kr',
        right_on='sigungu_2_kr',
        how='left'
    )
    merged = merged[merged['year'] <= merged['base_year']]
    return merged


def assign_adm2_code_by_alt_code(post_df: pd.DataFrame, lookup: pd.DataFrame, src_col: str, lookup_col: str) -> pd.DataFrame:
    merged = post_df.merge(
        lookup,
        left_on=src_col,
        right_on=lookup_col,
        how='left'
    )
    merged = merged[merged['year'] <= merged['base_year']]
    return merged
```
