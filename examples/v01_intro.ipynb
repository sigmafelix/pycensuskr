# Working with pycensuskr

```python
from pycensuskr import CensusKR
import pandas as pd
import geopandas as gpd

census = CensusKR()
```

## 1. Understanding Korean Geographic Hierarchies

South Korean census data is usually summarized at:
- `adm1`: Si-Do (province-level)
- `adm2`: Si-Gun-Gu (district-level)

`pycensuskr` provides district boundaries and tabular census data aligned by administrative codes.

The main census years are 2010, 2015, and 2020. ADM3 population and housing data is also available for 2015, 2020, and 2025.


```python
# Load 2020 district boundaries and inspect the number of adm2 units
adm2_2020 = census.load_districts(2020)
print('Number of adm2 districts in 2020:', adm2_2020['adm2_code'].nunique())
adm2_2020.head()
```

## 2. Available Census Data

The package includes a long-format dataset internally, and `anycensus()` returns analysis-friendly wide data.

Coverage varies by year and geographic level. ADM3 population and housing records cover 2015, 2020, and 2025.

Typical data types include:
- population
- housing
- tax
- mortality
- economy
- medicine
- migration
- environment
- social security
- welfare


```python
# Query district-level mortality data
mortality_adm2 = census.anycensus(year=2020, type='mortality', level='adm2')
mortality_adm2.head()
```

```python
# Aggregate to adm1 (province) level using mean
mortality_adm1 = census.anycensus(
    year=2020,
    type='mortality',
    level='adm1',
    aggregator='mean'
)
mortality_adm1.head()
```

```python
# Inspect the underlying long-format data table
long_2020 = census.load_data(2020)
long_2020.head()
```
