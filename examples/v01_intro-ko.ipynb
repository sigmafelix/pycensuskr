# pycensuskr 시작하기


```python
from pycensuskr import CensusKR
import pandas as pd
import geopandas as gpd

census = CensusKR()
```

## 1. 한국 행정구역 체계

주요 단위:
- `adm1`: 시도
- `adm2`: 시군구

`pycensuskr`는 행정코드를 기준으로 경계도형과 통계표를 함께 제공합니다.

주요 센서스 연도는 2010년, 2015년, 2020년입니다. 읍면동 단위 인구 및 주택 자료는 2015년, 2020년, 2025년도 제공됩니다.


```python
adm2_2020 = census.load_districts(2020)
print('2020년 시군구 수:', adm2_2020['adm2_code'].nunique())
adm2_2020.head()
```

## 2. 가용 센서스 데이터

`anycensus()`는 조회 결과를 분석하기 쉬운 wide 포맷으로 반환합니다.

자료 범위는 연도와 행정구역 단위에 따라 다릅니다. 읍면동 단위 인구 및 주택 자료는 2015년, 2020년, 2025년에 제공됩니다.


```python
df_2020 = census.anycensus(year=2020, type='mortality', level='adm2')
df_2020.head()
```

```python
df_2020_sido = census.anycensus(
    year=2020,
    type='mortality',
    level='adm1',
    aggregator='mean'
)
df_2020_sido.head()
```
